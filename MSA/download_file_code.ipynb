```python
import requests

# URL of the raw file
url = 'http://phylogeny.lirmm.fr/phylo_cgi/get_result.cgi?task_id=c60ec89a3cade760b237e57c0b79416a&results_in_list=2&raw=1&file=align_cured.phy'

# Send a GET request to the URL
response = requests.get(url)

# Save the content as a .phy file
with open('output_file.phy', 'wb') as file:
    file.write(response.content)

print("File downloaded successfully!")
```
